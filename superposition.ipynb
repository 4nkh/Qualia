# 1. Simuler la "Superposition" (L'intuition/Photosynthèse)

### Ici, on crée un qubit qui représente une décision ou un chemin. Au lieu d'être 0 ou 1, il explore les deux grâce à la porte de Hadamard ($H$).

```python
import pennylane as qml
from pennylane import numpy as np

# On crée un simulateur avec 1 qubit
dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit_superposition():
    # La porte Hadamard met le qubit en superposition
    # C'est comme l'énergie de la photosynthèse qui prend tous les chemins
    qml.Hadamard(wires=0)
    return qml.probs(wires=0)

print(f"Probabilités (Chemin A, Chemin B) : {circuit_superposition()}")
```

# 2. Simuler la "Normalisation Associative" (L'Intrication)

### On peut simuler l'idée que deux événements (ou un sens et un qualia) deviennent indissociables. Si l'un change, l'autre réagit instantanément, peu importe la distance. C'est l'intrication.

```python
dev2 = qml.device("default.qubit", wires=2)

@qml.qnode(dev2)
def circuit_intrication():
    # Événement 1 (le signal sensoriel)
    qml.Hadamard(wires=0)
    
    # Événement 2 (le Qualia) est lié à l'Événement 1
    # Le CNOT crée l'intrication (le "normalisateur")
    qml.CNOT(wires=[0, 1])
    
    return qml.state()

print(f"État quantique intriqué : {circuit_intrication()}")
```
