# On va construire un petit Classificateur Quantique Variationnel (VQC). 

### L'idée est de simuler ta théorie : comment des signaux bruts (tes perceptions) sont "normalisés" vers un état spécifique (le Qualia).

### On va utiliser un qubit comme une "boussole" : selon le signal d'entrée, le circuit va faire tourner ce qubit pour qu'il pointe vers le "Nord" (Qualia A) ou le "Sud" (Qualia B).

```python
import pennylane as qml
from pennylane import numpy as np

# 1. On définit le simulateur (1 qubit = notre espace de conscience)
dev = qml.device("default.qubit", wires=1)

# 2. Le circuit "Apprenant"
@qml.qnode(dev)
def qualia_circuit(inputs, weights):
    # ENTRÉE : On encode le signal sensoriel (ex: fréquence lumineuse)
    # On fait tourner le qubit selon la valeur du signal
    qml.RX(inputs[0], wires=0)
    
    # NORMALISATION : Le circuit utilise des "poids" (ton vécu/instinct)
    # pour ajuster la perception
    qml.RY(weights[0], wires=0)
    
    # On mesure la probabilité d'être dans l'état "Qualia"
    return qml.expval(qml.PauliZ(0))

# 3. Test avec des données
signal_brut = np.array([0.5])  # Un signal sensoriel quelconque
poids_instinctifs = np.array([1.2]) # Tes associations apprises

resultat = qualia_circuit(signal_brut, poids_instinctifs)

print(f"Interprétation du Qualia (entre -1 et 1) : {resultat}")
```

# Ce que ce test simule techniquement :

### qml.RX (Le Signal) : C'est la donnée brute qui frappe tes sens. En physique quantique, cela crée une rotation sur la Sphère de Bloch.

### qml.RY (Le Poids/L'Instinct) : C'est là que ta "normalisation" intervient. Ces poids représentent comment ton cerveau a été entraîné par l'évolution et ton expérience pour dévier le signal vers une émotion ou une catégorie précise.

### qml.PauliZ (La Mesure) : C'est l'effondrement de la fonction d'onde. On sort du monde des probabilités pour obtenir une sensation nette : "C'est du bleu" ou "C'est un danger".
